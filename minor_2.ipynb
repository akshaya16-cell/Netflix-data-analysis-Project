```python
df["Category_Mean"] = (
    df.groupby("category")["Amount"]
      .transform("mean")
)
```

```python
df["Category_Std"] = (
    df.groupby("category")["Amount"]
      .transform("std")
)
```

```python
df["z_score"] = (
    (df["Amount"] - df["Category_Mean"])
    / df["Category_Std"]
)
```

```python
df[[
    "Description",
    "category",
    "Amount",
    "z_score"
]].head(10)
```

```python
anomalies = df[df["z_score"] > 3]
```

```python
print(anomalies[[
    "Date",
    "Description",
    "category",
    "Amount",
    "z_score"
]])
```

```python
anomalies = anomalies.sort_values(
    by="z_score",
    ascending=False
)

print(anomalies)
```
