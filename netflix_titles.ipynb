```python
#checking null values
df.isnull().sum()
```

```python
#handling the nnull values
df["director"]=df["director"].fillna("unknown")
df["cast"]=df["cast"].fillna("unknown")
df["country"]=df["country"].fillna("NA")
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
```

```python
df = df.dropna(subset=["date_added"])
df = df.dropna(subset=["duration"])
```

```python
df.isnull().sum()
```

```python
df.duplicated()
```

```python
plt.bar(type_counts.index,type_counts.values,color=["red","pink"])
plt.title("TYPE")
plt.xlabel("content type")
plt.ylabel("number of titles")
plt.show()
```

Movies account for the majority of Netflix's content library, indicating that the platform has historically focused more on films than TV shows.

```python
data = df.groupby("release_year")["show_id"].count()
import matplotlib.pyplot as plt

data = df.groupby("release_year")["show_id"].count()

plt.figure(figsize=(12,6))
plt.plot(data.index, data.values, marker="o")

plt.title("Netflix Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles Released")

plt.grid(True)
plt.show()
```

Content production increased rapidly after 2015, reflecting Netflix's expansion into original content.

```python
print(data.idxmax())   # Year
print(data.max())      # Number of titles
```

```python
data=df["country"].value_counts().head(10)
plt.barh(data.index,data.values)
plt.title("top 10 country ")
plt.xlabel("number of title")
plt.ylabel("country")
plt.show()
```

```python
data=df["rating"].value_counts()
plt.figure(figsize=(10,5))
bars=plt.bar(data.index,data.values)
plt.title("Distribution of netflix ratings")
plt.xlabel("ratings")
plt.ylabel("number of titles")
plt.xticks(rotation=45)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        int(bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.show()
```

TV-MA is the most common rating, suggesting that Netflix primarily targets mature audiences.

```python
df["director"] = df["director"].replace("unkonwn", "Unknown")
```

```python
data = df[df["director"] != "Unknown"]["director"].value_counts().head(10)
plt.barh(data.index,data.values)
plt.title("top 10 directors ")
plt.xlabel("number of title")
plt.ylabel("directors")
plt.show()
```

```python
data = (
    df[df["cast"] != "unkonwn"]["cast"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
plt.barh(data.index, data.values)

plt.title("Top 10 Actors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Actors")

plt.show()
```

```python
movies = df[df["type"] == "Movie"].copy()
tvshows = df[df["type"] == "TV Show"].copy()
```

```python
movies["duration"] = movies["duration"].str.replace(" min", "").astype(int)
```

```python
print("Average:", movies["duration"].mean())
print("Longest:", movies["duration"].max())
print("Shortest:", movies["duration"].min())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(movies["duration"], bins=20)

plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

```python
tvshows["duration"]=tvshows["duration"].str.extract(r'(\d+)').astype(int)
```

```python
print("Average:", tvshows["duration"].mean())
print("Longest:", tvshows["duration"].max())
print("Shortest:", tvshows["duration"].min())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(tvshows["duration"], bins=20)

plt.title("Distribution of tvshows Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")

plt.show()
```

```python
genres = df["listed_in"].str.split(", ").explode()

top_genres = genres.value_counts().head(10)

import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.barh(top_genres.index, top_genres.values)

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")
plt.ylabel("Genre")

plt.show()
```

international Movies and Dramas dominate the catalog, indicating strong investment in globally appealing content.
